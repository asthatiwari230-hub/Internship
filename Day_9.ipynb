```python
from sklearn.neighbors import KNeighborsClassifier

# Features
X = [
    [2],
    [3],
    [5],
    [6],
    [7]
]

# Labels
y = ["Fail", "Fail", "Pass", "Pass", "Pass"]

# Create model
model = KNeighborsClassifier(n_neighbors=3)

# Train
model.fit(X, y)

# Predict
prediction = model.predict([[4]])

print(prediction)
```
