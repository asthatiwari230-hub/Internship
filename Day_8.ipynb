```python
from sklearn.linear_model import LinearRegression
X=[[1],[2],[3],[4],[5],[6],[7]]
Y=[[1000],[1200],[1500],[1800],[5000],[7563],[6473]]
model=LinearRegression()
model.fit(X,Y)
prediction=model.predict([[7]])
print("Predicted Salary:",prediction[0])
print(model.coef_)
print(model.intercept_)
```
